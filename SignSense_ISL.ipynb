```python
from ultralytics import YOLO
import cv2
```

```python
model = YOLO("yolov8n.pt")
```

```python
model.train(
    data="ISL_dataset/data.yaml",
    epochs=25,
    imgsz=640
)
```

```python
model = YOLO("runs/detect/train/weights/best.pt")
```

```python
import os
os.listdir("ISL_dataset/test/images")
```

```python
import os
import matplotlib.pyplot as plt

img = os.listdir("ISL_dataset/test/images")[0]

results = model.predict(
    source=f"ISL_dataset/test/images/{img}",
    show=False,
    save=False
)

annotated = results[0].plot()

plt.imshow(annotated)
plt.axis("off")
plt.show()
```
